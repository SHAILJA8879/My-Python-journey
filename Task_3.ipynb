```python
# ============================================
# TASK 3: NLP - SENTIMENT ANALYSIS
# ============================================

import string
import nltk

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB

# Download NLTK tokenizer
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

# --------------------------------------------
# 1. Dataset - 12 Product Reviews
# --------------------------------------------

reviews = [
    "I loved this product, amazing quality and great value",
    "Worst purchase ever, very disappointed and waste of money",
    "Excellent service and fast delivery, highly recommended",
    "The item broke after one day, terrible build quality",
    "Absolutely fantastic, exceeded my expectations",
    "Not worth the price, poor packaging and slow shipping",
    "Great product, will definitely buy again",
    "Stopped working in a week, very bad experience",
    "Beautiful design, works perfectly, very happy",
    "Useless, refund process is also a nightmare",
    "Quality is superb, packaging was secure",
    "Defective piece received, no response from seller"
]

labels = [
    "positive",
    "negative",
    "positive",
    "negative",
    "positive",
    "negative",
    "positive",
    "negative",
    "positive",
    "negative",
    "positive",
    "negative"
]

# --------------------------------------------
# 2. Text Preprocessing
# --------------------------------------------

def preprocess(text):
    # Convert text to lowercase
    text = text.lower()

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    return text


cleaned = [preprocess(review) for review in reviews]

# Tokenization using NLTK
tokens = [nltk.word_tokenize(review) for review in cleaned]

print("First tokenized review:")
print(tokens[0])

# --------------------------------------------
# 3. Bag of Words
# --------------------------------------------

cv = CountVectorizer()

bow = cv.fit_transform(cleaned)

print("\nBag of Words (BoW) shape:", bow.shape)

# --------------------------------------------
# 4. TF-IDF
# --------------------------------------------

tfidf = TfidfVectorizer()

X_tfidf = tfidf.fit_transform(cleaned)

print("TF-IDF shape:", X_tfidf.shape)

# --------------------------------------------
# 5. Train Multinomial Naive Bayes
# --------------------------------------------

model = MultinomialNB()

model.fit(X_tfidf, labels)

print("\nNaive Bayes classifier trained successfully!")

# --------------------------------------------
# 6. Predict Sentiment of New Reviews
# --------------------------------------------

new_reviews = [
    "The product is wonderful and very useful",
    "Completely useless, do not buy this product",
    "Quality is decent but delivery was very slow"
]

new_cleaned = [
    preprocess(review)
    for review in new_reviews
]

new_vec = tfidf.transform(new_cleaned)

predictions = model.predict(new_vec)

# Display predictions
print("\nPredictions:")

for review, prediction in zip(new_reviews, predictions):
    print(review, "->", prediction)
```
